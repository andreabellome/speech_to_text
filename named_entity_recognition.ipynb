```python
import spacy

activated = spacy.prefer_gpu( gpu_id=0 ) # use GPU if available
nlp = spacy.load("it_core_news_lg") # load the model
```

```python
# define the categories to recognize
ner_categories = ["PERSON", "ORG", "GPE", "PRODUCT"]

# tokenize the text
text = "Robert Downey Jr. is an American actor and producer. He is best known for his roles in films such as Iron Man, The Avengers, and Sherlock Holmes. Downey has won several awards for his acting, including two Screen Actors Guild Awards and a Golden Globe Award. He has also been nominated for an Academy Award."
text_it = "Luca Bianchi è nato a Milano il 15 marzo 1985. Dopo aver completato gli studi in Ingegneria Informatica presso il Politecnico di Torino, ha lavorato per diverse aziende, tra cui Enel e TIM. Attualmente lavora come Data Scientist a Roma, e nel 2022 ha pubblicato un libro intitolato 'Machine Learning per Tutti' con la casa editrice Mondadori."

doc = nlp(text_it)
```

```python
# identify and classify named entities
entities = []
for ent in doc.ents:
    if ent.label_ in ner_categories:
        entities.append((ent.text, ent.label_))

# print the named eneities and their categories
for entity, category in entities:
    print(f"{entity}: {category}")
```

```python
spacy.displacy.render(doc, style="ent")
```
